```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from pathlib import Path

DATA = Path("../data/processed")

RF = 0.065
TRADING_DAYS = 252
```

```python
nav = pd.read_csv(DATA / "nav_history_clean.csv")
```

```python
print(nav.shape)
nav.head()
```

```python
nav["date"] = pd.to_datetime(nav["date"])
```

```python
nav["daily_return"] = (
    nav.groupby("amfi_code")["nav"]
       .pct_change()
)
```

```python
sharpe_df = nav.groupby("amfi_code")["daily_return"].agg(
    mean_return="mean",
    std_return="std"
)

sharpe_df["sharpe_ratio"] = (
    (sharpe_df["mean_return"] * TRADING_DAYS - RF)
    /
    (sharpe_df["std_return"] * np.sqrt(TRADING_DAYS))
)

sharpe_df[["sharpe_ratio"]].head()
```

```python
benchmark = pd.read_csv(
    DATA / "benchmark_indices_clean.csv"
)

benchmark.head()
```

```python
print(benchmark.columns.tolist())
```

```python
benchmark.head()
```

```python
benchmark["date"] = pd.to_datetime(benchmark["date"])
```

```python
nifty100 = benchmark[
    benchmark["index_name"] == "NIFTY100"
].copy()
```

```python
nifty100["benchmark_return"] = (
    nifty100["close_value"]
    .pct_change()
)
```

```python
nifty100.head()
```

```python
nav["date"] = pd.to_datetime(nav["date"])
```

```python
from scipy.stats import linregress
```

```python
alpha_beta = []

for code, group in nav.groupby("amfi_code"):

    merged = pd.merge(
        group[["date", "daily_return"]],
        nifty100[["date", "benchmark_return"]],
        on="date",
        how="inner"
    ).dropna()

    if len(merged) < 30:
        continue

    slope, intercept, r, p, se = linregress(
        merged["benchmark_return"],
        merged["daily_return"]
    )

    alpha_beta.append([
        code,
        intercept * 252,
        slope
    ])

alpha_beta_df = pd.DataFrame(
    alpha_beta,
    columns=["amfi_code", "alpha", "beta"]
)

alpha_beta_df.head()
```

```python
alpha_beta_df.to_csv(
    REPORTS / "alpha_beta.csv",
    index=False
)
```

```python
mdd_list = []

for code, group in nav.groupby("amfi_code"):

    group = group.sort_values("date")

    running_max = group["nav"].cummax()

    drawdown = (
        group["nav"] / running_max
    ) - 1

    mdd = drawdown.min()

    mdd_list.append([
        code,
        mdd
    ])

mdd_df = pd.DataFrame(
    mdd_list,
    columns=["amfi_code", "max_drawdown"]
)

mdd_df.head()
```

```python
scorecard = (
    cagr_df[["cagr"]]
    .merge(
        sharpe_df[["sharpe_ratio"]],
        left_index=True,
        right_index=True
    )
    .merge(
        alpha_beta_df[["alpha"]],
        left_index=True,
        right_on="amfi_code"
    )
    .merge(
        mdd_df,
        left_index=True,
        right_on="amfi_code"
    )
)
```

```python
cagr_df = nav.groupby("amfi_code").agg(
    start_nav=("nav", "first"),
    end_nav=("nav", "last"),
    start_date=("date", "first"),
    end_date=("date", "last")
)

cagr_df["years"] = (
    cagr_df["end_date"] - cagr_df["start_date"]
).dt.days / 365

cagr_df["cagr"] = (
    (cagr_df["end_nav"] / cagr_df["start_nav"])
    ** (1 / cagr_df["years"])
    - 1
)

cagr_df[["cagr"]].head()
```

```python
sortino_list = []

for code, group in nav.groupby("amfi_code"):

    downside = group.loc[
        group["daily_return"] < 0,
        "daily_return"
    ]

    downside_std = downside.std()

    mean_return = group["daily_return"].mean()

    sortino = (
        (mean_return * TRADING_DAYS - RF)
        /
        (downside_std * np.sqrt(TRADING_DAYS))
    )

    sortino_list.append(
        [code, sortino]
    )

sortino_df = pd.DataFrame(
    sortino_list,
    columns=["amfi_code", "sortino_ratio"]
)

sortino_df.head()
```

```python
# Make sure dates are datetime
cagr_df["start_date"] = pd.to_datetime(cagr_df["start_date"])
cagr_df["end_date"] = pd.to_datetime(cagr_df["end_date"])

# Calculate years
cagr_df["years"] = (
    cagr_df["end_date"] - cagr_df["start_date"]
).dt.days / 365

# Calculate CAGR
cagr_df["cagr"] = (
    (cagr_df["end_nav"] / cagr_df["start_nav"])
    ** (1 / cagr_df["years"])
    - 1
)

cagr_df.head()
```

```python
scorecard = (
    cagr_df[["cagr"]]
    .join(sharpe_df[["sharpe_ratio"]])
)

scorecard.head()
```

```python
scorecard = scorecard.reset_index()

scorecard = scorecard.merge(
    alpha_beta_df[["amfi_code", "alpha"]],
    on="amfi_code",
    how="left"
)

scorecard = scorecard.merge(
    mdd_df[["amfi_code", "max_drawdown"]],
    on="amfi_code",
    how="left"
)

scorecard.head()
```

```python
scorecard["return_rank"] = scorecard["cagr"].rank(
    ascending=False
)

scorecard["sharpe_rank"] = scorecard["sharpe_ratio"].rank(
    ascending=False
)

scorecard["alpha_rank"] = scorecard["alpha"].rank(
    ascending=False
)

scorecard["dd_rank"] = scorecard["max_drawdown"].rank(
    ascending=False
)
```

```python
scorecard["score"] = (
      0.40 * scorecard["return_rank"]
    + 0.30 * scorecard["sharpe_rank"]
    + 0.20 * scorecard["alpha_rank"]
    + 0.10 * scorecard["dd_rank"]
)
```

```python
scorecard["score"] = (
    scorecard["score"]
    / scorecard["score"].max()
) * 100
```

```python
top5 = scorecard.sort_values(
    "score",
    ascending=False
).head(5)

top5
```

```python
scorecard.to_csv(
    REPORTS / "fund_scorecard.csv",
    index=False
)
```
